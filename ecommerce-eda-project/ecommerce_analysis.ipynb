```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

clean_data = pd.read_csv('cleaned_data1.csv')

# clean_data.head(5)

clean_data['TotalAmount'] = clean_data['Quantity'] * clean_data['UnitPrice']

clean_data.head(5)


# Find the top 5 countries by total revenue (TotalAmount).

clean_data.groupby('Country')['TotalAmount'].sum().sort_values(ascending=False).head(5)
```

```python
# Find all unique product descriptions (Description) bought by customers in
#  'Germany' where the single order line total (TotalAmount) was greater than 100.


unique_des = clean_data.loc[
    (clean_data['Country'] == 'Germany') & (clean_data['TotalAmount'] > 100),'Description'
].unique()
print(unique_des)




```

```python

# clean_data.head(5)

clean_data['InvoiceDate']


clean_data['InvoiceDate'] = pd.to_datetime(clean_data['InvoiceDate'])

# Step 2: Extract individual components into separate columns
clean_data['Year'] = clean_data['InvoiceDate'].dt.year
clean_data['Month'] = clean_data['InvoiceDate'].dt.month
clean_data['Day'] = clean_data['InvoiceDate'].dt.day
clean_data['DayName'] = clean_data['InvoiceDate'].dt.day_name() # e.g., 'Wednesday'
clean_data['Hour'] = clean_data['InvoiceDate'].dt.hour
clean_data['Minute'] = clean_data['InvoiceDate'].dt.minute
clean_data['Second'] = clean_data['InvoiceDate'].dt.second
```

```python

# Find which hour of the day (Hour) generates the highest average sales amount per order line (TotalAmount).

avg_sales = clean_data.groupby('Hour')['TotalAmount'].mean().sort_values(ascending=False)
print(avg_sales)

```

```python
# Select rows from index position 100 to 110 (inclusive) for only the columns 'InvoiceNo', 'Description', and 'TotalAmount'.

clean_data.loc[100:110,['InvoiceNo','Description','TotalAmount']]
```

```python
# Find the top 5 most frequently sold products that contain the word "HEART" in their description.

# df.info()

heart_items = clean_data[clean_data['Description'].str.contains('HEART', case=False, na=False)]
top5_heart = heart_items['Description'].value_counts().head(5)

print(top5_heart)
```

```python
# Create a summary table showing total sales (TotalAmount sum) broken down by DayOfWeek across the rows and Hour across the columns

pd.pivot_table(
    clean_data, 
    values='TotalAmount',  # The numerical column to sum up
    index='DayOfWeek',     # Rows of the matrix
    columns='Hour',        # Columns of the matrix
    aggfunc='sum',         # Aggregation function
    fill_value=0           # Fill NaN values with 0
)

```

```python
# Find the top 5 products (by Description) that generated the highest total revenue (TotalAmount), but only for orders placed in the month of 'December'
# . Display the product description and its total revenue sorted in descending order.


clean_data[clean_data['Month'] == 'December'].groupby('Description')['TotalAmount'].sum().sort_values(ascending=False).head(5)


```

```python
# Create a new column called Order_Category based on Quantity:
# If Quantity is 1 to 10 $\rightarrow$ 'Small'
# If Quantity is 11 to 50 $\rightarrow$ 'Medium'
# If Quantity is greater than 50 $\rightarrow$ 'Bulk'
# Then, calculate the total revenue (TotalAmount sum) generated by each Order_Category.

conditions = [
    (clean_data['Quantity'] >= 1) & (clean_data['Quantity'] <= 10),
    (clean_data['Quantity'] >= 11) & (clean_data['Quantity'] <= 50),
    (clean_data['Quantity'] > 50)
]

choice = ['Small','Medium','Bulk']

clean_data['Order_Category'] = np.select(conditions,choice,default='other')
clean_data.groupby('Order_Category')['TotalAmount'].sum().sort_values(ascending=False)

```

```python
# Plot a line chart showing the total sales revenue (TotalAmount) by Month across the year.   


monthly_sales = clean_data.groupby('Month')['TotalAmount'].sum().reset_index()

sns.lineplot(x='Month', y='TotalAmount', data=monthly_sales, marker='o')
plt.title("Total Sales Revenue")
plt.xlabel('Month')
plt.ylabel('Total Revenue ($)')
plt.xticks(range(1,13))
plt.grid(True)
plt.show()


```

```python
# Identify the Top 10 Countries by Total Revenue (excluding 'United Kingdom').
# Requirements:

# Filter out 'United Kingdom' from your dataset.

# Calculate total revenue (TotalAmount sum) for each remaining country and select the top 10.

# Plot a horizontal or vertical bar plot (plt.bar() or sns.barplot()).

# Include clean axis labels and a title.


total_rev = clean_data[
    (clean_data['Country'] != 'United Kingdom')
].groupby('Country')['TotalAmount'].sum().sort_values(ascending=False).head(10).reset_index()


plt.figure(figsize=(10, 5))
sns.barplot(x='Country', y='TotalAmount', data=total_rev, palette='viridis')

# 3. Rotate x labels so country names don't overlap
plt.xticks(rotation=45)
plt.title("Top 10 International Countries by Total Revenue (Excl. UK)")
plt.ylabel("Total Revenue ($)")
plt.tight_layout()

plt.show()



```

```python
# Create a heatmap showing total revenue (TotalAmount sum) across DayOfWeek (rows) and Hour (columns).

# Create the pivot table using 'DayName'
pivot = pd.pivot_table(
    clean_data,
    values='TotalAmount',  # Column to sum
    index='DayName',       # Use DayName for ROWS
    columns='Hour',        # Use Hour for COLUMNS
    aggfunc='sum',
    fill_value=0
)

# Reorder days chronologically
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Sunday']
pivot = pivot.reindex(days_order).dropna(how='all')

# Plot the heatmap
plt.figure(figsize=(10, 5))
sns.heatmap(pivot, cmap='YlGnBu')
plt.title("Heatmap of Revenue by Day of Week and Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Day of Week")
plt.show()
```

```python
# Find the Top 10 Best-Selling Products by TotalAmount and plot them as a horizontal bar chart.

top_products = clean_data.groupby('Description')['TotalAmount'].sum().sort_values(ascending=False).head(10).reset_index()



sns.barplot(x='TotalAmount',y='Description',data=top_products,palette='viridis')

plt.title("Top 10 Best-Selling Products by Revenue")
plt.ylabel("Total Revenue ($)")
plt.show()





```

```python
# Plot a histogram with a KDE density curve showing how Quantity is distributed across customer orders.

# Requirements:

# Filter clean_data to orders where Quantity is between 1 and 100 (to focus on standard orders and ignore extreme bulk outliers).

# Use sns.histplot(x='Quantity', data=..., kde=True, bins=30, color='teal').

# Add a title ("Distribution of Order Quantities (Filtered <= 100)"), label the x-axis ("Quantity per Line Item"), and call plt.show().


filtered_data = clean_data[(clean_data['Quantity'] >= 1) & (clean_data['Quantity'] <= 100)]
plt.figure(figsize=(10, 5))
sns.histplot(x='Quantity', data=filtered_data, kde=True, bins=30, color='teal')


plt.title("Distribution of Order Quantities (Filtered <= 100)")
plt.xlabel("Quantity per Line Item")
plt.ylabel("Frequency")
plt.tight_layout()

plt.show()
```
